```python
import numpy as np
import matplotlib.pyplot as plt

# Valores fornecidos
valores_x = [40, 22, 31, 33, 24, 34, 30]
valores_y = [28, 26, 25, 28, 26, 33, 27]

# Cria o gráfico de dispersão para visualizar a correlação
plt.figure(figsize=(8, 6))
plt.scatter(valores_x, valores_y, color='blue', label='Pontos de Dados')

# Adiciona títulos e rótulos dos eixos
plt.title('Gráfico de Correlação Modelo x Julia')
plt.xlabel('Valores Julia')
plt.ylabel('Valores Modelo')

# Adiciona linha de tendência (regressão linear)
m, b = np.polyfit(valores_x, valores_y, 1)  # Encontra a linha de regressão
plt.plot(valores_x, [m*x + b for x in valores_x], color='red', label='Linha de Tendência')

# Adiciona legenda
plt.legend()

# Mostra o gráfico
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Valores fornecidos
valores_x = [40, 22, 31, 33, 24, 34, 30]
valores_y = [28, 26, 25, 28, 26, 33, 27]

# Calcula o coeficiente de correlação de Pearson
correlacao_pearson, _ = pearsonr(valores_x, valores_y)

# Cria o gráfico de dispersão
plt.figure(figsize=(8, 6))
plt.scatter(valores_x, valores_y, color='blue', label=f'Pontos de Dados (r={correlacao_pearson:.2f})')

# Adiciona títulos e rótulos dos eixos
plt.title('Gráfico de Correlação de Pearson')
plt.xlabel('Valores Observados')
plt.ylabel('Valores Modelo')

# Adiciona linha de tendência (regressão linear)
m, b = np.polyfit(valores_x, valores_y, 1)
plt.plot(valores_x, [m*x + b for x in valores_x], color='red', label='Linha de Tendência')

# Adiciona legenda
plt.legend()

# Mostra o gráfico
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Valores fornecidos
valores_x = [40, 22, 31, 33, 24, 34, 30]
valores_y = [28, 26, 25, 28, 26, 39, 27]

# Calcula o coeficiente de correlação de Pearson
correlacao_pearson, _ = pearsonr(valores_x, valores_y)

# Cálculo dos resíduos (diferença entre valores observados e preditos) por ponto
residuos = np.array(valores_x) - np.array(valores_y)

# Calcula a média dos resíduos e a apresenta em %
media_residuos = np.mean(residuos)
media_residuos_percentual = (media_residuos / np.mean(valores_x)) * 100

# Cria o gráfico de dispersão
plt.figure(figsize=(8, 6))
plt.scatter(valores_x, valores_y, color='blue', label=f'Pontos de Dados (r={correlacao_pearson:.2f})')

# Adiciona títulos e rótulos dos eixos
plt.title(f'Gráfico de Correlação de Pearson e Resíduos\nMédia dos Resíduos: {media_residuos_percentual:.2f}%')
plt.xlabel('Valores Observados (X)')
plt.ylabel('Valores Preditos (Y)')

# Define a escala do eixo Y de 25 a 50
plt.ylim(25, 50)

# Adiciona linha de tendência (regressão linear)
m, b = np.polyfit(valores_x, valores_y, 1)
plt.plot(valores_x, [m*x + b for x in valores_x], color='red', label='Linha de Tendência')

# Adiciona os resíduos por ponto no gráfico
for i in range(len(valores_x)):
    plt.annotate(f'Res: {residuos[i]:.2f}', (valores_x[i], valores_y[i]), textcoords="offset points", xytext=(0,5), ha='center')

# Adiciona legenda
plt.legend()

# Mostra o gráfico
plt.grid(True)
plt.show()

# Exibir os resíduos e a média dos resíduos
print(f'Resíduos (Valores Observados - Valores Preditos) por ponto: {residuos}')
print(f'Média dos Resíduos: {media_residuos:.2f}')
print(f'Média dos Resíduos em %: {media_residuos_percentual:.2f}%')
```
