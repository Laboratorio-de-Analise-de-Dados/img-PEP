```python
#gerar gráfico
import matplotlib
#estruturação dos dados
import numpy as np
#gerar gráfico
import seaborn as sns
#gerar gráfico
import matplotlib.pyplot as plt
#Carregar o modelo
from keras.models import load_model
#salvar/carregar arquivos em diferentes formatos
import joblib
```

```python
#Importando o modelo contendo os valores previstos para o número de células nas imagens
modelo_maior_erro = load_model('/content/drive/MyDrive/Projeto PEP Julia/teste/imagens_julia_quantificacao/lista de imagens e rótulos - contraste digital/img_e_rot_para_validação_contraste/imagens rotulo modelo - validação campo claro/modelo_campo_claro_100_75_50_25_2024-9-10.keras')
modelo_maior_erro
```

```python
X_test_maior_erro = joblib.load('/content/drive/MyDrive/Projeto PEP Julia/teste/imagens_BAELB/img.gz_campo_claro/lista_img_100_resized_regre_campo_claro_2025-2-11.gz') #carregando arquivo
X_test_maior_erro.shape
```

```python
# #ROTULOS SALVOS EM LISTA

# lista_observado_maior_erro = list(y_test_maior_erro)
# len(lista_observado_maior_erro)
```

```python
#PREDIÇÃO SALVO EM LISTA

dados_prev = modelo_maior_erro.predict(X_test_maior_erro)
lista_previsto_maior_erro = dados_prev.flatten().tolist()
len(lista_previsto_maior_erro)
```

```python
# Visualizar previsões
plt.figure(figsize=(10, 5))
plt.plot(lista_previsto_maior_erro, marker='o', linestyle='')
plt.xlabel("Amostras")
plt.ylabel("Valores Previstos")
plt.title("Visualização dos Valores Preditos")
plt.show()
```

```python
# Exibir uma imagem com o valor predito
indice_amostra = 28  # Escolher um índice de amostra para visualizar
imagem_exibida = X_test_maior_erro[indice_amostra]
valor_predito = lista_previsto_maior_erro[indice_amostra]

plt.figure(figsize=(5, 5))
plt.imshow(imagem_exibida.squeeze(), cmap='gray')  # Certifica-se de que a imagem está em escala de cinza
plt.title(f"Valor Predito: {valor_predito}")
plt.axis("off")
plt.show()
```
